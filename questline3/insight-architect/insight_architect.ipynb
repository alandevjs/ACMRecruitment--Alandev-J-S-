```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("cleaned_students_performance.csv")
```

```python
#Histogram
#How are students' math scores spread from 0-100

plt.hist(df["math score"])
plt.title("Distribution of math scores")
plt.xlabel("Maths score")
plt.ylabel("Number of students")
plt.show()
```

```python
#Barchart
#Do male and female students have different average scores

avg_math=df.groupby("gender")["math score"].mean()
plt.bar(avg_math.index, avg_math.values)

plt.title("Average math score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average math scores")
plt.show()
```

```python
#boxplot
#How math scores differ between students who completed the test preparation course and those who didn't.

sns.boxplot(x="test preparation course", y="math score", data=df)

plt.title("Math scores by Test Preparation Course")
plt.xlabel("Test Preparation course")
plt.ylabel("Math score")
plt.show()
```

```python
#scatterplot
#examinign relationship btwn math score and reading score
plt.scatter(df["math score"], df["reading score"])

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
#heatmap
#finding correlation between math score, reading score and writing score

corr = df[["math score", "reading score", "writing score"]].corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Between Student Scores")
plt.show()
```

# #Key Insights
#Most of the students scored between 50 and 80 in math, with many scores around 60–70.
#The male students had a higher average math score than the female students.
#Students who completed the test preparation course had a higher median math score than students who did not complete it.
#The scatter plot shows that students with higher math scores usually also had higher reading scores.
#Math, reading, and writing scores all have a positive relationship, meaning students who do well in one subject often do well in the other subjects too.
#Reading and writing had the strongest relationship, with a correlation of 0.95.
